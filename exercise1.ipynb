```python
def string_operations(s):
    rev_str = s[::-1]
    second_char = s[1::2]
    return rev_str, second_char

print(string_operations("Artificial Intelligence"))
```

```python
def square_evens(nums):
    return [x**2 for x in nums if x % 2 == 0]

print(square_evens([1, 2, 3, 4, 5, 6]))
```

```python
def char_frequency(s):
    freq = {}
    for char in s:
        freq[char] = freq.get(char, 0) + 1
    return freq

print(char_frequency("hello ai"))
```

```python
def lists_to_dict(list1, list2):
    return dict(zip(list1, list2))

print(lists_to_dict(['a', 'b', 'c'], [1, 2, 3]))
```

```python
def manual_iter_example(lst):
    it = iter(lst)
    while True:
        try:
            print(next(it))
        except StopIteration:
            print("End of list reached!")
            break

manual_iter_example([10, 20, 30])
```

```python
def is_palindrome(s):
    return s == s[::-1]

print(is_palindrome("madam"))
```

```python
def second_largest(nums):
    unique_nums = list(set(nums))
    if len(unique_nums) < 2:
        return None
    first = second = float('-inf')
    for n in unique_nums:
        if n > first:
            second = first
            first = n
        elif n > second:
            second = n
    return second

print(second_largest([10, 20, 4, 45, 99]))
```
